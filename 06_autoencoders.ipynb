The model should learn to remove the noise, since it is random in nature and have no relationship with the images.
  d. Anomaly Detection
  - the purpose of anomaly detection is to reduce reconstruction error. Generally speaking, this can be done with the following steps:
  1. train the encoder on the whole dataset, assuming outliers are negligible. If outliers are present, select a portion of the dataset that does not contain outliers.
  2. Run the outlier inputs through the model and calculate the RE of those outputs
  3. Sort the observations by RE values
  4. Classify the observations with the highest RE as outliers. This requires an analysis of the results and a lot of knowledge about the data and the problem.

4. An FFA model and a PCA will be the same if the following conditions are met:
- a linear function is used for the encoder g()

If those conditions are not met, then the autoencoder has a computational advantage on PCA due to it being able to deal with a very big amount of data efficently since its training can be done in mini-batches, whereas PCA has to use the entire dataset. Additionally, autoencoders are more flexible and they consider non-linear transformations of the features, whereas the default PCA method uses strictly linear and cannot scale up with the increasing data size, making it a non-practical solution from a computational point of view.
